```python
# NORTH–SOUTH GEOGRAPHIC SPREAD
# Full violin: spreads LEFT + RIGHT

import plotly.graph_objects as go

BRAND_COLORS = {
    "Blinkit": "#F8CB46",
    "Zepto": "#7A288A",
    "Swiggy Instamart": "#FC8019"
}

brand_order = [
    "Blinkit",
    "Zepto",
    "Swiggy Instamart"
]

fig = go.Figure()

for brand in brand_order:

    data = stores.loc[
        stores["brand"] == brand,
        "latitude"
    ].dropna()

    fig.add_trace(
        go.Violin(
            x=[brand] * len(data),
            y=data,

            orientation="v",

            # FULL violin — no side="positive"
            name=brand,

            line=dict(
                color=BRAND_COLORS[brand],
                width=2.2
            ),

            fillcolor=BRAND_COLORS[brand],
            opacity=0.60,

            box_visible=False,
            meanline_visible=False,
            points=False,

            width=0.75,

            hovertemplate=(
                f"<b>{brand}</b><br>"
                "Latitude: %{y:.2f}°N"
                "<extra></extra>"
            )
        )
    )

fig.update_layout(

    title=dict(
        text=(
            "<b>North–South Geographic Spread</b><br>"
            "<span style='font-size:13px;color:#777777'>"
            "Latitude concentration of each quick-commerce network"
            "</span>"
        ),
        x=0.5,
        xanchor="center",
        font=dict(size=23, color="#202124")
    ),

    xaxis=dict(
        title="",
        showgrid=False
    ),

    yaxis=dict(
        title="SOUTH  ↓   Latitude (°N)   ↑  NORTH",
        gridcolor="#EEEEEE",
        zeroline=False
    ),

    showlegend=False,

    plot_bgcolor="#FCFCFC",
    paper_bgcolor="white",

    width=1000,
    height=650,

    margin=dict(
        t=125,
        l=110,
        r=60,
        b=80
    )
)

fig.show()
```

```python
# EAST–WEST GEOGRAPHIC SPREAD
# Full violin: spreads UP + DOWN

import plotly.graph_objects as go

BRAND_COLORS = {
    "Blinkit": "#F8CB46",
    "Zepto": "#7A288A",
    "Swiggy Instamart": "#FC8019"
}

brand_order = [
    "Blinkit",
    "Zepto",
    "Swiggy Instamart"
]

fig = go.Figure()

for brand in brand_order:

    data = stores.loc[
        stores["brand"] == brand,
        "longitude"
    ].dropna()

    fig.add_trace(
        go.Violin(
            x=data,
            y=[brand] * len(data),

            orientation="h",

            # FULL violin — no side="positive"
            name=brand,

            line=dict(
                color=BRAND_COLORS[brand],
                width=2.2
            ),

            fillcolor=BRAND_COLORS[brand],
            opacity=0.60,

            box_visible=False,
            meanline_visible=False,
            points=False,

            width=0.75,

            hovertemplate=(
                f"<b>{brand}</b><br>"
                "Longitude: %{x:.2f}°E"
                "<extra></extra>"
            )
        )
    )

fig.update_layout(

    title=dict(
        text=(
            "<b>East–West Geographic Spread</b><br>"
            "<span style='font-size:13px;color:#777777'>"
            "Longitude concentration of each quick-commerce network"
            "</span>"
        ),
        x=0.5,
        xanchor="center",
        font=dict(size=23, color="#202124")
    ),

    xaxis=dict(
        title="WEST  ←   Longitude (°E)   →  EAST",
        gridcolor="#EEEEEE",
        zeroline=False
    ),

    yaxis=dict(
        title="",
        showgrid=False
    ),

    showlegend=False,

    plot_bgcolor="#FCFCFC",
    paper_bgcolor="white",

    width=1000,
    height=550,

    margin=dict(
        t=125,
        l=150,
        r=60,
        b=80
    )
)

fig.show()
```

```python
import plotly.graph_objects as go

BRAND_COLORS = {
    "Blinkit": "#F8CB46",
    "Zepto": "#7A288A",
    "Swiggy Instamart": "#FC8019"
}

total_stores = brand_counts.sum()

fig = go.Figure()

for brand, count in brand_counts.items():

    share = count / total_stores * 100

    fig.add_trace(
        go.Bar(
            x=[brand],
            y=[count],
            marker=dict(
                color=BRAND_COLORS[brand],
                line=dict(color="white", width=2)
            ),
            text=f"<b>{count:,}</b><br>{share:.1f}%",
            textposition="outside",
            hovertemplate=(
                f"<b>{brand}</b><br>"
                f"Stores: {count:,}<br>"
                f"Network Share: {share:.1f}%"
                "<extra></extra>"
            ),
            showlegend=False
        )
    )

fig.update_layout(
    title=dict(
        text=(
            "<b>QUICK-COMMERCE NETWORK SCALE</b><br>"
            "<sup>Dark store distribution across competing brands</sup>"
        ),
        x=0.5
    ),

    xaxis_title="Brand",
    yaxis_title="Number of Dark Stores",

    plot_bgcolor="#FAFAFA",
    paper_bgcolor="white",

    yaxis=dict(
        gridcolor="#EAEAEA",
        zeroline=False
    ),

    height=520,
    margin=dict(t=110, l=80, r=50, b=70)
)

fig.show()
```

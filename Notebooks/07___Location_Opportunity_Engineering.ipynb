```python
# Load intelligence layers

demand = pd.read_csv("retail_demand_index.csv")

competition = pd.read_csv(
    "store_competition_features (1).csv"
)

osm = pd.read_csv(
    "osm_proof_of_concept (2).csv"
)

print("Demand:", demand.shape)
print("Competition:", competition.shape)
print("OSM:", osm.shape)
```

```python
# Check available features

print("\nDEMAND")
print(demand.columns.tolist())

print("\nCOMPETITION")
print(competition.columns.tolist())

print("\nOSM")
print(osm.columns.tolist())
```

```python
# Demand features

demand_features = demand[
    [
        "State",
        "District",
        "Name",
        "No_HH",
        "TOT_P",
        "Literacy_Rate",
        "Workforce_Rate",
        "Average_Household_Size",
        "Retail_Demand_Index"
    ]
].copy()
```

```python
# Competition features

competition_features = competition[
    [
        "id",
        "brand",
        "latitude",
        "longitude",
        "city",
        "state",
        "location_lat",
        "location_lon",
        "Total_Dark_Stores",
        "Competing_Brands",
        "Competition_Intensity",
        "Same_Brand_Stores",
        "Competitor_Stores",
        "Competitor_Ratio",
        "Brand_Share"
    ]
].copy()
```

```python
# Environment features

osm_features = osm[
    [
        "brand",
        "latitude",
        "longitude",
        "nearby_schools",
        "nearby_colleges",
        "nearby_hospitals",
        "nearby_malls",
        "public_transport_points",
        "commercial_areas",
        "residential_areas",
        "neighborhood_activity_score"
    ]
].copy()
```

```python
# Check layer coverage

print("Demand records:", len(demand_features))
print("Competition records:", len(competition_features))
print("OSM records:", len(osm_features))

print("\nOSM brand coverage:")
print(osm_features["brand"].value_counts())
```

```python
# Check geographic fields

print("Competition cities:")
print(
    competition_features["city"]
    .dropna()
    .value_counts()
    .head(20)
)

print("\nDemand markets:")
print(
    demand_features["Name"]
    .dropna()
    .value_counts()
    .head(20)
)
```

```python
# Standardize market names

competition_features["market_key"] = (
    competition_features["city"]
    .astype(str)
    .str.strip()
    .str.lower()
)

demand_features["market_key"] = (
    demand_features["Name"]
    .astype(str)
    .str.strip()
    .str.lower()
)
```

```python
# Check demand and competition overlap

competition_markets = set(
    competition_features["market_key"]
)

demand_markets = set(
    demand_features["market_key"]
)

matched = competition_markets & demand_markets

print("Competition markets:", len(competition_markets))
print("Demand markets:", len(demand_markets))
print("Matched markets:", len(matched))

print("\nMatched names:")
print(sorted(matched))
```

```python
# Fix common city-name differences

name_map = {
    "bengaluru": "bangalore",
    "gurugram": "gurgaon",
    "sas nagar": "mohali"
}

competition_features["market_key"] = (
    competition_features["market_key"]
    .replace(name_map)
)
```

```python
# Recheck market coverage

competition_markets = set(
    competition_features["market_key"]
)

demand_markets = set(
    demand_features["market_key"]
)

matched = competition_markets & demand_markets

print("Competition markets:", len(competition_markets))
print("Matched markets:", len(matched))

print("\nStill unmatched:")
print(
    sorted(
        competition_markets - demand_markets
    )
)
```

```python
# Create market-level demand layer

market_demand = (
    demand_features
    .dropna(subset=["market_key", "Retail_Demand_Index"])
    .groupby("market_key", as_index=False)
    .agg(
        Population=("TOT_P", "max"),
        Households=("No_HH", "max"),
        Literacy_Rate=("Literacy_Rate", "mean"),
        Workforce_Rate=("Workforce_Rate", "mean"),
        Average_Household_Size=("Average_Household_Size", "mean"),
        Retail_Demand_Index=("Retail_Demand_Index", "max")
    )
)

print("Market demand records:", len(market_demand))

market_demand.head()
```

```python
# Merge demand with store competition

opportunity = competition_features.merge(
    market_demand,
    on="market_key",
    how="inner"
)

print("Opportunity dataset:", opportunity.shape)

print("\nMarkets represented:")
print(opportunity["market_key"].nunique())

print("\nBrands:")
print(opportunity["brand"].value_counts())

opportunity.head()
```

```python
# Check brands at each stage

print("Original competition:")
print(competition["brand"].value_counts(dropna=False))

print("\nCompetition features:")
print(competition_features["brand"].value_counts(dropna=False))

print("\nMatched opportunity:")
print(opportunity["brand"].value_counts(dropna=False))
```

```python
# Check market coverage by brand

brand_market_check = (
    competition_features
    .groupby("brand")["market_key"]
    .nunique()
    .sort_values(ascending=False)
)

print(brand_market_check)
```

```python
# Check missing city values by brand

print(
    competition
    .groupby("brand")["city"]
    .agg(
        total="size",
        city_available="count"
    )
)
```

```python
# Create coordinate-based market keys

competition_features["geo_lat"] = (
    competition_features["latitude"]
    .round(1)
)

competition_features["geo_lon"] = (
    competition_features["longitude"]
    .round(1)
)

competition_features["geo_key"] = (
    competition_features["geo_lat"].astype(str)
    + "_"
    + competition_features["geo_lon"].astype(str)
)

print(
    competition_features[
        ["brand", "latitude", "longitude", "geo_key"]
    ].head()
)

print("\nStores retained:")
print(len(competition_features))

print("\nBrand coverage:")
print(competition_features["brand"].value_counts())
```

```python
# Build market coordinate reference

market_reference = (
    competition_features[
        competition_features["city"].notna()
    ]
    .groupby("market_key", as_index=False)
    .agg(
        market_lat=("latitude", "median"),
        market_lon=("longitude", "median")
    )
)

print("Reference markets:", len(market_reference))

market_reference.head()
```

```python
# Connect reference markets with demand

market_reference = market_reference.merge(
    market_demand,
    on="market_key",
    how="inner"
)

print("Markets with demand:", len(market_reference))

market_reference[
    [
        "market_key",
        "market_lat",
        "market_lon",
        "Retail_Demand_Index"
    ]
].head()
```

```python
# Calculate geographic distance

def haversine(lat1, lon1, lat2, lon2):

    R = 6371

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    return 2 * R * np.arcsin(np.sqrt(a))
```

```python
# Assign nearest demand market

def nearest_market(row):

    distances = haversine(
        row["latitude"],
        row["longitude"],
        market_reference["market_lat"].values,
        market_reference["market_lon"].values
    )

    idx = np.argmin(distances)

    return pd.Series({
        "assigned_market":
            market_reference.iloc[idx]["market_key"],

        "market_distance_km":
            distances[idx]
    })


assigned = competition_features.apply(
    nearest_market,
    axis=1
)

competition_features = pd.concat(
    [
        competition_features.reset_index(drop=True),
        assigned.reset_index(drop=True)
    ],
    axis=1
)
```

```python
# Validate nearest-market assignment

print(
    competition_features[
        "market_distance_km"
    ].describe()
)

print("\nDistance bands:")

print(
    pd.cut(
        competition_features["market_distance_km"],
        bins=[0, 10, 25, 50, 100, 250, np.inf],
        labels=[
            "0-10 km",
            "10-25 km",
            "25-50 km",
            "50-100 km",
            "100-250 km",
            "250+ km"
        ]
    ).value_counts().sort_index()
)
```

```python
# Classify demand assignment quality

competition_features["assignment_quality"] = np.select(
    [
        competition_features["market_distance_km"] <= 25,
        competition_features["market_distance_km"] <= 50
    ],
    [
        "High Confidence",
        "Moderate Confidence"
    ],
    default="Outside Market Range"
)

print(
    competition_features["assignment_quality"]
    .value_counts()
)
```

```python
# Keep stores within 50 km

opportunity = competition_features[
    competition_features["market_distance_km"] <= 50
].copy()

print("Stores available for opportunity modelling:", len(opportunity))

print("\nBrand coverage:")
print(opportunity["brand"].value_counts())

print("\nMarkets represented:")
print(opportunity["assigned_market"].nunique())
```

```python
# Attach market demand features

demand_lookup = market_reference[
    [
        "market_key",
        "Population",
        "Households",
        "Literacy_Rate",
        "Workforce_Rate",
        "Average_Household_Size",
        "Retail_Demand_Index"
    ]
].rename(
    columns={"market_key": "assigned_market"}
)

opportunity = opportunity.merge(
    demand_lookup,
    on="assigned_market",
    how="left"
)

print("Final opportunity layer:", opportunity.shape)

print("\nMissing RDI:")
print(opportunity["Retail_Demand_Index"].isna().sum())

print("\nBrand coverage:")
print(opportunity["brand"].value_counts())
```

```python
# Normalize opportunity features

def minmax(series):
    return (
        (series - series.min()) /
        (series.max() - series.min())
    )

opportunity["Demand_Strength"] = minmax(
    opportunity["Retail_Demand_Index"]
)

opportunity["Competition_Risk"] = minmax(
    opportunity["Competition_Intensity"]
)

opportunity["Competitor_Pressure"] = minmax(
    opportunity["Competitor_Ratio"]
)

opportunity["Local_Brand_Strength"] = minmax(
    opportunity["Brand_Share"]
)

opportunity["Market_Density"] = minmax(
    opportunity["Total_Dark_Stores"]
)
```

```python
# Create opportunity signals

opportunity["Demand_Gap"] = (
    opportunity["Demand_Strength"]
    * (1 - opportunity["Competition_Risk"])
)

opportunity["Expansion_Headroom"] = (
    opportunity["Demand_Strength"]
    * (1 - opportunity["Competitor_Pressure"])
)

opportunity["Market_Control"] = (
    opportunity["Local_Brand_Strength"]
    * opportunity["Demand_Strength"]
)

opportunity[
    [
        "brand",
        "assigned_market",
        "Demand_Strength",
        "Competition_Risk",
        "Demand_Gap",
        "Expansion_Headroom",
        "Market_Control"
    ]
].head()
```

```python
# Demand vs competition opportunity matrix

import plotly.express as px

plot_df = (
    opportunity
    .groupby("assigned_market", as_index=False)
    .agg(
        Demand_Strength=("Demand_Strength", "mean"),
        Competition_Risk=("Competition_Risk", "mean"),
        Expansion_Headroom=("Expansion_Headroom", "mean"),
        Dark_Stores=("Total_Dark_Stores", "mean"),
        Population=("Population", "mean")
    )
)

demand_mid = plot_df["Demand_Strength"].median()
risk_mid = plot_df["Competition_Risk"].median()

fig = px.scatter(
    plot_df,

    x="Competition_Risk",
    y="Demand_Strength",

    size="Dark_Stores",
    color="Expansion_Headroom",

    hover_name="assigned_market",

    hover_data={
        "Population": ":,.0f",
        "Dark_Stores": ":.0f",
        "Demand_Strength": ":.2f",
        "Competition_Risk": ":.2f",
        "Expansion_Headroom": ":.2f"
    },

    color_continuous_scale=[
        "#D8EDE8",
        "#55B6A9",
        "#2878A5",
        "#313695"
    ],

    size_max=45
)

# Quadrants

fig.add_vline(
    x=risk_mid,
    line_dash="dash",
    line_color="#AAAAAA"
)

fig.add_hline(
    y=demand_mid,
    line_dash="dash",
    line_color="#AAAAAA"
)

# Opportunity zone

fig.add_annotation(
    x=0.03,
    y=0.97,
    xref="paper",
    yref="paper",

    text=(
        "<b>EXPANSION SWEET SPOT</b><br>"
        "High Demand • Lower Competition"
    ),

    showarrow=False,

    bgcolor="rgba(85,182,169,0.12)",
    bordercolor="#55B6A9",
    borderpad=7
)

# Saturated zone

fig.add_annotation(
    x=0.97,
    y=0.97,
    xref="paper",
    yref="paper",

    text=(
        "<b>HIGH-VALUE BATTLEGROUND</b><br>"
        "High Demand • High Competition"
    ),

    showarrow=False,

    bgcolor="rgba(49,54,149,0.08)",
    bordercolor="#313695",
    borderpad=7
)

fig.update_traces(
    marker=dict(
        opacity=0.75,
        line=dict(
            color="white",
            width=1.5
        )
    )
)

fig.update_layout(

    title=dict(
        text=(
            "<b>WHERE DOES DEMAND OUTRUN COMPETITION?</b><br>"
            "<span style='font-size:13px;color:#777777'>"
            "Market opportunity emerges where demand remains strong "
            "without excessive competitive pressure"
            "</span>"
        ),
        x=0.5,
        xanchor="center",
        font=dict(
            size=22,
            color="#263B5E"
        )
    ),

    xaxis=dict(
        title="Competitive Risk →",
        gridcolor="#EEEEEE",
        zeroline=False
    ),

    yaxis=dict(
        title="Demand Strength →",
        gridcolor="#EEEEEE",
        zeroline=False
    ),

    coloraxis_colorbar=dict(
        title="Expansion<br>Headroom"
    ),

    plot_bgcolor="#FCFCFC",
    paper_bgcolor="white",

    width=1000,
    height=650,

    margin=dict(
        t=125,
        l=90,
        r=120,
        b=80
    )
)

fig.show()
```

```python
# Rank markets by expansion headroom

top_opportunities = (
    plot_df
    .sort_values(
        "Expansion_Headroom",
        ascending=False
    )
    .head(12)
    .sort_values("Expansion_Headroom")
)

# Create ranking chart

fig = px.bar(
    top_opportunities,

    x="Expansion_Headroom",
    y="assigned_market",

    orientation="h",

    color="Demand_Strength",

    color_continuous_scale=[
        "#B8E0D2",
        "#55B6A9",
        "#2878A5",
        "#313695"
    ],

    text="Expansion_Headroom",

    hover_data={
        "Demand_Strength": ":.2f",
        "Competition_Risk": ":.2f",
        "Population": ":,.0f",
        "Dark_Stores": ":.0f"
    }
)

fig.update_traces(
    texttemplate="%{text:.2f}",
    textposition="outside"
)

fig.update_layout(

    title=dict(
        text=(
            "<b>MARKETS WITH THE STRONGEST EXPANSION HEADROOM</b><br>"
            "<span style='font-size:13px;color:#777777'>"
            "High demand combined with manageable competitor pressure"
            "</span>"
        ),
        x=0.5,
        xanchor="center",
        font=dict(
            size=22,
            color="#263B5E"
        )
    ),

    xaxis=dict(
        title="Expansion Headroom →",
        gridcolor="#EEEEEE",
        range=[
            0,
            top_opportunities[
                "Expansion_Headroom"
            ].max() * 1.12
        ]
    ),

    yaxis=dict(
        title="",
        tickfont=dict(size=12)
    ),

    coloraxis_colorbar=dict(
        title="Demand<br>Strength"
    ),

    plot_bgcolor="#FCFCFC",
    paper_bgcolor="white",

    width=1000,
    height=620,

    margin=dict(
        t=120,
        l=150,
        r=130,
        b=70
    )
)

fig.show()
```

```python
blinkit = pd.read_csv("blinkit-darkstores.csv")
zepto = pd.read_csv("zepto-darkstores.csv")
swiggy = pd.read_csv("swiggy-darkstores.csv")
```

```python
print("Blinkit")
display(blinkit.head())

print("Zepto")
display(zepto.head())

print("Swiggy")
display(swiggy.head())
```

```python
print("Blinkit")
blinkit.info()

print("\nZepto")
zepto.info()

print("\nSwiggy")
swiggy.info()
```

```python
blinkit["brand"] = "Blinkit"
zepto["brand"] = "Zepto"
swiggy["brand"] = "Swiggy Instamart"
```

```python
rename_columns = {
    "lat": "latitude",
    "lng": "longitude"
}

blinkit.rename(columns=rename_columns, inplace=True)
zepto.rename(columns=rename_columns, inplace=True)
swiggy.rename(columns=rename_columns, inplace=True)
```

```python
# Blinkit
blinkit["locality"] = np.nan
blinkit["city"] = np.nan
blinkit["state"] = np.nan

# Swiggy
swiggy["city"] = np.nan
swiggy["state"] = np.nan

# Zepto already has:
# name, city, state

# Rename 'name' to 'locality' so all datasets match
zepto.rename(columns={"name": "locality"}, inplace=True)
```

```python
columns = [
    "id",
    "locality",
    "city",
    "state",
    "latitude",
    "longitude",
    "brand"
]

stores = pd.concat(
    [
        blinkit[columns],
        zepto[columns],
        swiggy[columns]
    ],
    ignore_index=True
)
```

```python
print("Shape:", stores.shape)

print("\nMissing Values:")
print(stores.isnull().sum())

print("\nDuplicate Rows:", stores.duplicated().sum())

print("\nBrand Distribution:")
print(stores["brand"].value_counts())
```

```python
stores = stores.drop_duplicates()
```

```python
stores.to_csv("master_darkstores.csv", index=False)

print("✅ master_darkstores.csv created successfully!")
```

```python
# Visualize Brand Network Share

import plotly.graph_objects as go

brand_counts = (
    stores["brand"]
    .value_counts()
    .reset_index()
)

brand_counts.columns = ["Brand", "Stores"]

brand_colors = {
    "Blinkit": "#FFD000",
    "Zepto": "#7B2CBF",
    "Swiggy Instamart": "#FC8019"
}

fig = go.Figure(
    data=[
        go.Pie(
            labels=brand_counts["Brand"],
            values=brand_counts["Stores"],

            hole=0.58,

            marker=dict(
                colors=[
                    brand_colors[x]
                    for x in brand_counts["Brand"]
                ],
                line=dict(
                    color="white",
                    width=3
                )
            ),

            textinfo="label+percent",

            hovertemplate=(
                "<b>%{label}</b><br>"
                "Dark Stores: %{value:,}<br>"
                "Network Share: %{percent}"
                "<extra></extra>"
            )
        )
    ]
)

fig.update_layout(
    title={
        "text": "Quick-Commerce Dark Store Network Share",
        "x": 0.5
    },

    annotations=[
        dict(
            text="<b>4,081</b><br>Total Stores",
            x=0.5,
            y=0.5,
            showarrow=False,
            font=dict(size=18)
        )
    ],

    template="plotly_white",
    height=600,

    legend=dict(
        orientation="h",
        x=0.5,
        xanchor="center",
        y=-0.05
    )
)

fig.show()
```

```python
# Visualize Geographic Dark Store Network

import plotly.express as px

brand_colors = {
    "Blinkit": "#FFD000",
    "Zepto": "#7B2CBF",
    "Swiggy Instamart": "#FC8019"
}

fig = px.scatter_map(
    stores,

    lat="latitude",
    lon="longitude",

    color="brand",

    color_discrete_map=brand_colors,

    hover_name="brand",

    hover_data={
        "latitude": ":.4f",
        "longitude": ":.4f"
    },

    title="Geographic Footprint of Quick-Commerce Dark Stores",

    center={
        "lat": 22.5,
        "lon": 79
    },

    zoom=3.5,
    height=750
)

fig.update_traces(
    marker=dict(
        size=6,
        opacity=0.55
    )
)

fig.update_layout(
    map_style="carto-positron",

    legend=dict(
        title="Brand",
        orientation="h",
        x=0.5,
        xanchor="center",
        y=1.02
    ),

    margin=dict(
        l=0,
        r=0,
        t=90,
        b=0
    )
)

fig.show()
```

```python
# Analyze Geographic Spread by Brand

fig = px.scatter(
    stores,

    x="longitude",
    y="latitude",

    color="brand",

    color_discrete_map={
        "Blinkit": "#FFD000",
        "Zepto": "#7B2CBF",
        "Swiggy Instamart": "#FC8019"
    },

    hover_name="brand",

    hover_data={
        "latitude": ":.4f",
        "longitude": ":.4f"
    },

    marginal_x="histogram",
    marginal_y="histogram",

    title="Spatial Distribution and Geographic Density by Brand",

    labels={
        "longitude": "Longitude",
        "latitude": "Latitude",
        "brand": "Brand"
    },

    height=750
)

fig.update_traces(
    marker=dict(
        opacity=0.45
    ),
    selector=dict(type="scatter")
)

fig.update_layout(
    template="plotly_white",

    legend=dict(
        title="Brand",
        orientation="h",
        x=0.5,
        xanchor="center",
        y=1.08
    )
)

fig.show()
```

```python
# Create Brand-Wise Geographic Footprint Maps

import plotly.express as px
from IPython.display import display, HTML

brand_config = {
    "Blinkit": {
        "color": "#FFD000",
        "text_color": "#111111"
    },
    "Zepto": {
        "color": "#7B2CBF",
        "text_color": "#FFFFFF"
    },
    "Swiggy Instamart": {
        "color": "#FC8019",
        "text_color": "#FFFFFF"
    }
}

for brand, config in brand_config.items():

    brand_data = stores[
        stores["brand"] == brand
    ].copy()

    # Brand-style heading
    display(
        HTML(
            f"""
            <div style="
                background:{config['color']};
                color:{config['text_color']};
                padding:12px 25px;
                border-radius:12px;
                display:inline-block;
                font-size:24px;
                font-weight:700;
                margin:20px 0 10px 0;
            ">
                {brand}
            </div>
            """
        )
    )

    fig = px.scatter_map(
        brand_data,

        lat="latitude",
        lon="longitude",

        hover_data={
            "latitude": ":.4f",
            "longitude": ":.4f",
            "brand": False
        },

        center={
            "lat": 22.5,
            "lon": 79
        },

        zoom=3.4,
        height=650
    )

    fig.update_traces(
        marker=dict(
            size=6,
            color=config["color"],
            opacity=0.72
        )
    )

    fig.update_layout(
        map_style="carto-positron",

        title={
            "text": f"{brand} Dark Store Geographic Footprint",
            "x": 0.5,
            "xanchor": "center"
        },

        margin=dict(
            l=0,
            r=0,
            t=70,
            b=0
        )
    )

    fig.show()
```
